```python
import pandas as pd
from Bio import SeqIO
from bioservices.uniprot import UniProt
from tqdm import tqdm
import ete3
from ete3 import Tree
from ete3 import NodeStyle
```

```python
u = UniProt(verbose=False)
```

```python
tree = Tree('TBE_named_tree.new')

ids = []

for leaf in tree:
    a = leaf.name.split(' ')[-1]
    ids.append(a)
```

```python
linige = []
for item in tqdm(ids):
    lineage = u.search(item, columns='lineage')
    print(item)
    print(lineage)
    finige = f"{item}:{lineage}" 
    linige.append(finige)
```

```python
kingdom1 = []
kingdom2 = []
together = []


for items in linige:
    if "(domain)" in items:
        first = items.split(' (domain)')[0]
        second = first.split(' ')[-1]
        third = first.split(':')[0]

                    
        kingdom1.append(third)
        kingdom2.append(second)
        together.append(f"{third}: {second}")

dicking = dict(zip(kingdom1, kingdom2))
```

```python
with open('kingdom.txt', 'w') as fp:
    for ite in together:
        # write each item on a new line
        fp.write("%s\n" % ite)
    print('Done')
```

```python
id = []
king = []

for a in together:
    b = a.split(':')[0]
    id.append(b)
    c = a.split(' ')[1]
    king.append(c)
```

```python
df = pd.DataFrame()
```

```python
df['id'] = id
```

```python
df['kingdom'] = king
```

```python
df = df.set_index('id')
```
