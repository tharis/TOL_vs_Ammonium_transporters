```python
from Bio import SeqIO
import re
```

```python
rec = []

for record in SeqIO.parse("PII_protein.fasta", "fasta"):
    rec.append(record.description)
```

```python
rec
```

```python
protein_names = []
for entry in rec:
    match = re.search(r'\|\S+\s+(.*?)\s+OS=', entry)
    if match:
        protein_names.append(match.group(1))

print(protein_names)
```

```python
protein_namesset = set(protein_names)
```
