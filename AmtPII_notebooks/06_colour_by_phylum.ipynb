```python
from bioservices.uniprot import UniProt
from Bio import SeqIO
from tqdm import tqdm
import ete3
from ete3 import Tree
from ete3 import NodeStyle
```

```python
u = UniProt(verbose=False)
```

```python
tree = Tree("TBE_named_tree_nit_noeuk.new")
```

```python
ids = []

for leaf in tree:
    ids.append(leaf.name.split(' ')[-1])
```

```python
ids
```

```python
fool = []
for item in tqdm(ids):
    lineage = u.search(item, columns='lineage')
    print(item)
    print(lineage)
    finige = f"{item}:{lineage}" 
    fool.append(finige)
```

```python
with open('ids.txt', 'w') as fp:
    for ite in ids:
        # write each item on a new line
        fp.write("%s\n" % ite)
    print('Done')
```

```python
with open('full_taxonomy.txt', 'w') as fp:
    for ite in fool:
        # write each item on a new line
        fp.write("%s\n" % ite)
    print('Done')
```

```python
with open('full_taxonomy.txt', 'r') as file:

    fool = []
    for line in file:
        line = line.strip()
        fool.append(line)

print(fool)
```

```python
phylum1 = []
phylum2 = []
together = []

for items in fool:
    if "(phylum)" in items:
        first = items.split(' (phylum)')[0]
        second = first.split(' ')[-1]
        third = first.split(':')[0]
        # print(f"items {items}")
        # print(f"First {first}")
        # print(f"Second {second}")
        # print(f"Third {third}")
        phylum1.append(third)
        phylum2.append(second)
        together.append(f"{third}: {second}")

dicphyl = dict(zip(phylum1, phylum2))
```

```python
# colouring only the fungal part to see what the evolution in the group is

Act = []
Bac = []
Elu = []
Met = []
Spi = []

for item in together:
    if item.split(' ')[1] == 'Actinomycetota':
        Act.append(item.split(':')[0])
    elif item.split(' ')[1] == 'Bacillota':
        Bac.append(item.split(':')[0])
    elif item.split(' ')[1] == 'Elusimicrobiota':
        Elu.append(item.split(':')[0])
    elif item.split(' ')[1] == 'Methanobacteriota':
        Met.append(item.split(':')[0])
    elif item.split(' ')[1] == 'Spirochaetota':
        Spi.append(item.split(':')[0])
```

```python
R = tree.get_midpoint_outgroup()
tree.set_outgroup(R)
```

```python
def conditional_function(node):
        for node in tree:
            if node.name.split(' ')[-1] in Act:
                nst1 = NodeStyle()
                nst1['bgcolor'] = 'Moccasin'
                node.set_style(nst1)
            elif node.name.split(' ')[-1] in Bac:
                nst3 = NodeStyle()
                nst3['bgcolor'] = 'Wheat'
                node.set_style(nst3)
            elif node.name.split(' ')[-1] in Elu:
                nst4 = NodeStyle()
                nst4['bgcolor'] = 'Peru'
                node.set_style(nst4)
            elif node.name.split(' ')[-1] in Met:
                nst3 = NodeStyle()
                nst3['bgcolor'] = 'Salmon'
                node.set_style(nst3)
            elif node.name.split(' ')[-1] in Spi:
                nst4 = NodeStyle()
                nst4['bgcolor'] = 'Cornsilk'
                node.set_style(nst4)
            elif node.name.split(' ')[-1] == "A0A9P1BFQ1":
                nst5 = NodeStyle()
                nst5['bgcolor'] = "ForestGreen"
                node.set_style(nst5)
```

```python
new_size = 2

def make_branches_bigger(node):
    node.img_style["hz_line_width"] = new_size # Change the horizotal lines stroke size
    node.img_style["vt_line_width"] = new_size # Change the vertical lines stroke size
    for c in node.children:
        make_branches_bigger(c)
```

```python
ts = ete3.TreeStyle()
ts.layout_fn = conditional_function, make_branches_bigger
ts.mode = "c"

ts.show_branch_support = True
ts.show_leaf_name = True

tree.render("figure_4.4.pdf", tree_style=ts)
```
